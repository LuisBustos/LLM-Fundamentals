```python
import torch #PyTorch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import time

#MODELO CORRIDO EN CPU

#Creamos una clase del modelo
class MLP(nn.Module):
    def __init__(self,input_dim,hidden_dim,output_dim):
        super(MLP,self).__init__()
        #Definición de capas ocultas
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, output_dim)

    def forward(self,x):
        x = F.relu(self.fc1(x)) #Función de activación
        x = self.fc2(x)
        return x

model = MLP(input_dim = 10, hidden_dim = 32, output_dim = 1) #Deginición de neuronas

criterion = nn.MSELoss() #Error cuadrático medio, para regresión
optimizer = optim.Adam(model.parameters(), lr = 0.01)  #lr: Learning Rate (Por debajo del 0)

x = torch.randn(200000,10)
y = torch.randn(200000,1)

start_time = time.time()
#fordward + Backward

model.train()
optimizer.zero_grad()
output = model(x)
loss = criterion(output, y)
loss.backward()
optimizer.step()
end_time = time.time()
elapsed = end_time - start_time

print('Perdida: ', loss.item())
print(f"Tiempo de entrenamiento: {elapsed:.6f} segundos")
```

```python
import torch #PyTorch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import time

#MODELO CORRIDO EN GPU

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Usando dispotivo: ',device)

#Creamos una clase del modelo
class MLP(nn.Module):
    def __init__(self,input_dim,hidden_dim,output_dim):
        super(MLP,self).__init__()
        #Definición de capas ocultas
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, output_dim)

    def forward(self,x):
        x = F.relu(self.fc1(x)) #Función de activación
        x = self.fc2(x)
        return x

model = MLP(input_dim = 10, hidden_dim = 32, output_dim = 1) #Deginición de neuronas

criterion = nn.MSELoss() #Error cuadrático medio, para regresión
optimizer = optim.Adam(model.parameters(), lr = 0.01)  #lr: Learning Rate (Por debajo del 0)

x = torch.randn(200000,10).to(device)
y = torch.randn(200000,1).to(device)

start_time = time.time()
#fordward + Backward

model.train()
optimizer.zero_grad()
output = model(x)
loss = criterion(output, y)
loss.backward()
optimizer.step()
end_time = time.time()
elapsed = end_time - start_time

print('Perdida: ', loss.item())
print(f"Tiempo de entrenamiento: {elapsed:.6f} segundos")
```
